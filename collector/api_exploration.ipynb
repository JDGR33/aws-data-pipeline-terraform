# Api exploration
First exploration of the APIs


```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("EIA_API_KEY")
if not API_KEY:
    raise RuntimeError("EIA_API_KEY is missing from the .env file")

URL = "https://api.eia.gov/v2/electricity/rto/fuel-type-data/data/"

params = {
    "api_key": API_KEY,
    "frequency": "hourly",
    "data[0]": "value",
    "facets[respondent][]": "CISO",  # California ISO
    "sort[0][column]": "period",
    "sort[0][direction]": "desc",
    "offset": 0,
    "length": 100
}

response = requests.get(URL, params=params)
response.raise_for_status()
data = response.json()

# Records reside inside data['response']['data']
records = data.get("response", {}).get("data", [])
print(f"Retrieved {len(records)} records.")
```

```python
URL = "https://api.eia.gov/v2/electricity/rto/region-data/data/"
params = {
    "api_key": API_KEY,
    "frequency": "hourly",
    "data[0]": "value",
    "facets[respondent][]": "CISO",       # California ISO
    "facets[type][]": "DF",              # Day-Ahead Forecast
    "start": "2026-08-01T00",
    "end": "2026-08-05T23",
    "sort[0][column]": "period",
    "sort[0][direction]": "desc",
    "length": 50
}

response = requests.get(URL, params=params)
data = response.json()

# Inspecting sample records
for row in data.get("response", {}).get("data", [])[:3]:
    print(f"Time: {row['period']} | Grid: {row['respondent-name']} | Forecast: {row['value']} {row['value-units']}")
```

```python
URL = "https://api.eia.gov/v2/electricity/retail-sales/data/"

params = {
    "api_key": API_KEY,
    "frequency": "monthly",
    "data[0]": "price",
    "data[1]": "sales",
    "data[2]": "revenue",
    "facets[stateid][]": "TX",            # Texas
    "facets[sectorid][]": "RES",          # Residential Sector
    "start": "2025-01",
    "sort[0][column]": "period",
    "sort[0][direction]": "desc",
    "length": 12
}

response = requests.get(URL, params=params)
data = response.json()

# Inspecting sample records
for row in data.get("response", {}).get("data", [])[:3]:
    print(f"Month: {row['period']} | State: {row['stateDescription']} | Avg Price: {row['price']} cents/kWh | Sales: {row['sales']} MWh")
```

```python
# Fetch hourly temperature & solar radiation for Houston, TX (29.76, -95.36)
url = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": 29.76,
    "longitude": -95.36,
    "hourly": "temperature_2m,direct_normal_irradiance,wind_speed_10m",
    "timezone": "America/Chicago"
}

response = requests.get(url, params=params)
data = response.json()

# Raw arrays inside data['hourly']
times = data["hourly"]["time"]
temps = data["hourly"]["temperature_2m"]
print(f"Retrieved {len(times)} hourly forecast points.")
```
